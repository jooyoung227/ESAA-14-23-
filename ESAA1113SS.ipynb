# 넷플릭스 주가 예측하기 RNN 코드 실습

## 2. 파이토치 데이터셋 구성하기

```python

import numpy as np
from torch.utils.data.dataset import Dataset

class Netflix(Dataset):  # 클래스 선언
    def __init__(self):
        # 데이터 읽기
        self.data = pd.read_csv("/netflix_train.csv")

        # 입력 데이터 정규화
        # 종가를 제외한 데이터
        # 0과 1 사이로 정규화
        self.data = self.data[["Open", "High", "Low", "Volume", "Close"]]
        self.data = self.data.astype("float32")

        self.data = (self.data - self.data.min()) / (self.data.max() - self.data.min())

        # 종가 데이터 정규화
        self.label = self.data["Close"].values
        self.data = self.data[["Open", "High", "Low", "Volume"]].values

    def __len__(self):
        return len(self.data) - 30  # 사용 가능한 배치 개수

    def __getitem__(self, i):
        # 입력 데이터 30일치 읽기
        data = self.data[i:i+30]

        # 종가 데이터 30일치 읽기
        label = self.label[i+30]

        return data, label
```

## 3. RNN 모델 정의

```python

# RNN 클래스 정의

import torch
import torch.nn as nn

class RNN(nn.Module):
    def __init__(self):
        super(RNN, self).__init__()

        # RNN층의 정의
        self.rnn = nn.RNN(input_size=4, hidden_size=8,
                          num_layers=2, batch_first=True)

        # 주가를 예측하는 MLP층 정의
        self.fc = nn.Linear(8, 1)

        # 활성화 함수 정의
        self.relu = nn.ReLU()

    # RNN 모델의 순전파 정의
    def forward(self, x, h0):
        # RNN층의 출력
        x, hn = self.rnn(x, h0)

        # MLP층의 입력으로 사용되게 모양 변경
        x = x[:, -1, :]

        # MLP층을 이용해 종가 예측
        x = self.fc(x)

        # 예측한 종가를 1차원 벡터로 표현
        x = x.view(-1)

        return x
```

## 4. 모델 학습하기

```python

# 모델과 데이터셋 정의
import tqdm
from torch.optim.adam import Adam
from torch.utils.data.dataloader import DataLoader

device = "cuda" if torch.cuda.is_available() else "cpu"

# 모델의 정의
model = RNN().to(device)

# 데이터셋의 정의
dataset = Netflix()
```

```python

# 데이터로더 정의
# 배치 크기를 32로 설정
loader = DataLoader(dataset, batch_size=32, shuffle=False)
```

```python

# 최적화 정의
# 학습률은 0.0001로 설정
# 사용할 최적화 설정
optim = Adam(model.parameters(), lr=0.0001)
```

```python

# 학습 루프 정의
# 에포크: 200
for epoch in range(200):
    iterator = tqdm.tqdm(loader)
    for data, label in iterator:
        optim.zero_grad()

        data = data.to(device)
        label = label.to(device)

        # 초기 은닉 상태
        h0 = torch.zeros(2, data.size(0), 8).to(device)

        # 모델의 예측값
        pred = model(data, h0)

        # 손실의 계산
        loss = nn.MSELoss()(pred, label)

        # 오차 역전파
        loss.backward()

        # 최적화 진행
        optim.step()

        iterator.set_description(f"epoch{epoch} loss: {loss.item()}")

# 모델 저장
torch.save(model.state_dict(), "rnn.pth")
```

## 5. 모델 성능 평가하기

```python

import matplotlib.pyplot as plt

# 예측값을 위한 데이터로더
loader = DataLoader(dataset, batch_size=32, shuffle=False)

# 예측값들을 저장하는 리스트
preds = []
total_loss = 0

with torch.no_grad():
    # 모델의 가중치 불러오기
    model.load_state_dict(torch.load("rnn.pth", map_location=device, weights_only=True))
    model.eval()

    for data, label in loader:
        data = data.to(device)
        label = label.to(device)

        # 초기 은닉 상태 정의
        h0 = torch.zeros(2, data.size(0), 8).to(device)

        # 모델의 예측값 출력
        pred = model(data, h0)

        # 예측값을 리스트에 추가
        preds += pred.cpu().numpy().tolist()

        # 손실 계산
        loss = nn.MSELoss()(pred, label)

        # 손실의 평균치 계산
        total_loss += loss.item() * len(data)

total_loss = total_loss / len(dataset)
total_loss
```

```python

# 그래프 출력
plt.plot(preds, label="prediction")
plt.plot(dataset.label[30:], label="actual")
plt.legend()
plt.show()
```
