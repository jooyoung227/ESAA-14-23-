```python
import os
import pandas as pd
```

1. 강남구

```python
base_path = '/content/drive/MyDrive/ESAA YB 4조/*** 최종 ***/강남구'
output_path = '/content/drive/MyDrive/ESAA YB 4조/*** 최종 ***/장서대출_강남구.csv'
target_columns = ['도서명', '저자', '출판사', '권', '주제분류번호', '도서권수', '대출건수']

file_list = []
for root, dirs, files in os.walk(base_path):
    for file in files:
        if file.lower().endswith('.csv') and '월별_통합본' not in file:
            file_list.append(os.path.join(root, file))

combined_list = []

if file_list:
    for file_path in file_list:
        try:
            try:
                df = pd.read_csv(file_path, encoding='utf-8')
            except:
                df = pd.read_csv(file_path, encoding='cp949')

            df.columns = df.columns.str.strip()
            available_columns = [col for col in target_columns if col in df.columns]
            combined_list.append(df[available_columns].copy())
        except:
            pass

    if combined_list:
        final_df = pd.concat(combined_list, ignore_index=True)
        existing_targets = [col for col in target_columns if col in final_df.columns]
        final_df = final_df[existing_targets]
        final_df.to_csv(output_path, index=False, encoding='utf-8-sig')

final_df.head()
```

# 행정동 단위 서울 생활인구 연령대 재분류
5세 단위로 제공된 연령별 생활인구 데이터를 10세 단위 연령대로 재분류하였다.

```python
from pathlib import Path
import pandas as pd
import csv
import re

INPUT_DIR = Path("/content/drive/MyDrive/ESAA YB 4조/주영/행정구역별 시간대 연령대 성별 유동인구")
OUTPUT_DIR = Path("/content/drive/MyDrive/ESAA YB 4조/*** 최종 ***/행정구역별 생활인구")

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

START_YM = "202306"
END_YM = "202605"

age_groups = {
    "0대": ["0세부터9세"],
    "10대": ["10세부터14세", "15세부터19세"],
    "20대": ["20세부터24세", "25세부터29세"],
    "30대": ["30세부터34세", "35세부터39세"],
    "40대": ["40세부터44세", "45세부터49세"],
    "50대": ["50세부터54세", "55세부터59세"],
    "60대": ["60세부터64세", "65세부터69세"],
    "70대이상": ["70세이상"],
}

base_cols = ["기준일ID", "시간대구분", "행정동코드", "총생활인구수"]


def get_yyyymm(file_path):
    match = re.search(r"(\d{6})", file_path.stem)
    return match.group(1) if match else None


def read_csv_safely(file_path):
    try:
        with open(file_path, encoding="utf-8-sig", newline="") as f:
            header = next(csv.reader(f))

        df = pd.read_csv(
            file_path,
            encoding="utf-8-sig",
            usecols=range(len(header)),
            dtype={
                "기준일ID": "string",
                "시간대구분": "string",
                "행정동코드": "string",
            },
            index_col=False,
        )

    except UnicodeDecodeError:
        with open(file_path, encoding="cp949", newline="") as f:
            header = next(csv.reader(f))

        df = pd.read_csv(
            file_path,
            encoding="cp949",
            usecols=range(len(header)),
            dtype={
                "기준일ID": "string",
                "시간대구분": "string",
                "행정동코드": "string",
            },
            index_col=False,
        )

    return df


def convert_5age_to_10age(df):

    out = df[base_cols].copy()

    for gender in ["남자", "여자"]:
        for age_label, parts in age_groups.items():
            cols = [f"{gender}{part}생활인구수" for part in parts]

            missing = [c for c in cols if c not in df.columns]
            if missing:
                raise KeyError(f"없는 컬럼이 있습니다: {missing}")

            out[f"{gender}{age_label}생활인구수"] = (
                df[cols]
                .apply(pd.to_numeric, errors="coerce")
                .sum(axis=1)
            )

    for age_label in age_groups.keys():
        out[f"{age_label}생활인구수"] = (
            out[f"남자{age_label}생활인구수"] +
            out[f"여자{age_label}생활인구수"]
        )

    return out
```

```python
files = sorted(INPUT_DIR.glob("LOCAL_PEOPLE_DONG_*.csv"))

print(f"찾은 CSV 파일 수: {len(files)}개")

processed_count = 0
skipped_count = 0
for_file_error = []

for file_path in files:
    yyyymm = get_yyyymm(file_path)


    if yyyymm is None:
        print(f"연월 확인 불가 → 건너뜀: {file_path.name}")
        skipped_count += 1
        continue


    if not (START_YM <= yyyymm <= END_YM):
        print(f"기간 밖 파일 → 건너뜀: {file_path.name}")
        skipped_count += 1
        continue

    print(f"처리 중: {file_path.name}")

    try:
        df = read_csv_safely(file_path)
        converted = convert_5age_to_10age(df)

        output_path = OUTPUT_DIR / f"{file_path.stem}_age10.csv"
        converted.to_csv(output_path, index=False, encoding="utf-8-sig")

        processed_count += 1
        print(f"저장 완료: {output_path.name}")

    except Exception as e:
        print(f"오류 발생: {file_path.name}")
        print(e)
        for_file_error.append(file_path.name)

print("=" * 50)
print("전체 변환 완료")
print(f"처리 완료 파일 수: {processed_count}개")
print(f"건너뛴 파일 수: {skipped_count}개")
print(f"오류 파일 수: {len(for_file_error)}개")
print(f"저장 위치: {OUTPUT_DIR}")

if for_file_error:
    print("오류 발생 파일 목록:")
    for name in for_file_error:
        print("-", name)
```

* IQR 기준으로 이상치를 탐지하였으나, 대출건수의 크거나 작은 값은 실제 수요를 반영한 값이므로 제거하지 않음

1. 강남구

```python
import pandas as pd
import numpy as np

input_path = '/content/drive/MyDrive/ESAA YB 4조/*** 최종 ***/장서대출_강남구.csv'
output_path = '/content/drive/MyDrive/ESAA YB 4조/*** 최종 ***/장서대출_강남구_전처리.csv'

target_columns = [
    '도서명',
    '저자',
    '출판사',
    '권',
    '주제분류번호',
    '도서권수',
    '대출건수'
]

try:
    df = pd.read_csv(input_path, encoding='utf-8')
except UnicodeDecodeError:
    df = pd.read_csv(input_path, encoding='cp949')

df.columns = df.columns.str.strip()

available_target = [c for c in target_columns if c in df.columns]
df = df[available_target].copy()

df = df.dropna(subset=['도서명'])
df = df[df['도서명'].astype(str).str.strip() != '']

fill_values = {
    '저자': '저자미상',
    '출판사': '출판사없음',
    '주제분류번호': '미분류'
}

for col, value in fill_values.items():
    if col in df.columns:
        df[col] = df[col].fillna(value).astype(str).str.strip()
        df[col] = df[col].replace('', value)

if '권' in df.columns:
    df['권'] = df['권'].astype(str).replace('nan', '').str.strip()

for col in ['도서권수', '대출건수']:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0).astype(int)
        df = df[df[col] >= 0]

# IQR 기반 이상치 탐지

for col in ['도서권수', '대출건수']:
    if col in df.columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        _ = df[(df[col] < lower) | (df[col] > upper)]

df = df.drop_duplicates()

df.to_csv(output_path, index=False, encoding='utf-8-sig')
```

```python
import pandas as pd
import glob
import os

folder = "/content/drive/MyDrive/ESAA YB 4조/*** 최종 ***/장서대출_전처리"

files = glob.glob(os.path.join(folder, "*.csv"))

kdc = {
    '0':'총류',
    '1':'철학',
    '2':'종교',
    '3':'사회과학',
    '4':'자연과학',
    '5':'기술과학',
    '6':'예술',
    '7':'언어',
    '8':'문학',
    '9':'역사',
    '미분류':'미분류'
}

result = []

for file in files:

    filename = os.path.basename(file)
    parts = filename.replace(".csv", "").split("_")
    gu = parts[1]

    try:
        df = pd.read_csv(
            file,
            dtype={"주제분류번호": str},
            low_memory=False
        )
    except pd.errors.EmptyDataError:
        print(f"{filename} -> 빈 파일이라 제외")
        continue

    df["주제분류번호"] = df["주제분류번호"].astype(str)

    df["대분류"] = df["주제분류번호"].apply(
        lambda x: "미분류" if x == "미분류" else x[0]
    )

    df["주제"] = df["대분류"].map(kdc)

    genre = df.groupby("주제")["대출건수"].sum()

    genre_ratio = (genre / genre.sum() * 100).round(2)

    row = {"자치구": gu}

    for subject, value in genre_ratio.items():
        row[subject] = value

    result.append(row)

final = pd.DataFrame(result)

final = final.fillna(0)

```

## 1. Priority Score: 이관 후보 도서 선별

Priority Score = 0.45 X 장기미대출비율 + 0.30 X (100 - 장서회전율정규화) + 0.25 X (100 - 미래대출지수정규화)





```python
import unicodedata
import pandas as pd
import os

folder_path = "/content/drive/MyDrive/ESAA YB 4조/*** 최종 ***/장서대출_전처리"

kdc_map = {
    0: "총류", 1: "철학", 2: "종교", 3: "사회과학", 4: "자연과학",
    5: "기술과학", 6: "예술", 7: "언어", 8: "문학", 9: "역사"
}

all_gu_list = []

for file in os.listdir(folder_path):
    if not (file.endswith(".xlsx") or file.endswith(".csv")):
        continue

    gu_name = file.replace(".csv", "").replace(".xlsx", "").split("_")[1]
    gu_name = unicodedata.normalize('NFC', gu_name)

    file_path = os.path.join(folder_path, file)
    if file.endswith(".xlsx"):
        df_temp = pd.read_excel(file_path)
    else:
        df_temp = pd.read_csv(file_path, low_memory=False)

    subj = df_temp["주제분류번호"].astype(str)
    is_valid = subj.str[0].str.isdigit()
    df_temp = df_temp[is_valid].copy()

    df_temp["KDC주제"] = (
        df_temp["주제분류번호"].astype(str).str[0].astype(int).map(kdc_map)
    )
    df_temp["자치구"] = gu_name

    all_gu_list.append(df_temp)

df_book_all = pd.concat(all_gu_list, ignore_index=True)
```

```python
df_priority = df_book_all.groupby(["자치구", "KDC주제"]).agg(
    총소장권수=("도서권수", "sum"),
    연간대출건수=("대출건수", "sum"),
    장기미대출권수=("대출건수", lambda x: (x == 0).sum())
).reset_index()
```

```python
#장기미대출비율
df_gu_summary = df_priority.groupby("자치구").agg(
    총소장권수=("총소장권수", "sum"),
    장기미대출권수=("장기미대출권수", "sum")
).reset_index()

df_gu_summary["장기미대출비율"] = (
    df_gu_summary["장기미대출권수"] /
    df_gu_summary["총소장권수"]
) * 100

df_gu_summary.sort_values("장기미대출비율", ascending=False)
```

```python
# 장서회전율 정규화
df_gu_summary = df_priority.groupby("자치구").agg(
    총소장권수=("총소장권수", "sum"),
    연간대출건수=("연간대출건수", "sum"),
    장기미대출권수=("장기미대출권수", "sum")
).reset_index()

df_gu_summary["장기미대출비율"] = (
    df_gu_summary["장기미대출권수"] /
    df_gu_summary["총소장권수"]
) * 100

df_gu_summary["장서회전율"] = (
    df_gu_summary["연간대출건수"] /
    df_gu_summary["총소장권수"]
)

min_turnover_gu = df_gu_summary["장서회전율"].min()
max_turnover_gu = df_gu_summary["장서회전율"].max()

df_gu_summary["장서회전율정규화"] = (
    (df_gu_summary["장서회전율"] - min_turnover_gu)
    / (max_turnover_gu - min_turnover_gu)
) * 100

df_gu_summary.sort_values("장서회전율", ascending=False)
```

##2. Potential Score: 자치구 별 도서 대출 가능도

y = 0.30×주제장서회전점수+0.20×주제선호도점수+0.20×생활인구점수+0.20×장서부족도점수+0.10×인구증가율점수

* Potential Score는 주제별 수요(장서회전율, 주제선호도)와 지역의 장서 확충 필요성(장서부족도, 생활인구, 인구증가율)을 함께 반영한 종합지표


###2-1. 주제 장서회전점수
Min-Max 정규화[log{1 + (해당 자치구의 해당 KDC 주제 대출건수 ÷ 해당 자치구의 해당 KDC 주제 총소장권수)}] × 100

```python
import pandas as pd
import numpy as np
import os
import unicodedata

loan_folder = (
    "/content/drive/MyDrive/ESAA YB 4조/*** 최종 ***/장서대출_전처리"
)

loan_files = [
    os.path.join(loan_folder, file_name)
    for file_name in os.listdir(loan_folder)
    if file_name.lower().endswith(".csv")
]

if len(loan_files) == 0:
    raise ValueError("장서대출 CSV 파일을 찾지 못했습니다.")

kdc_map = {
    "0": "총류",
    "1": "철학",
    "2": "종교",
    "3": "사회과학",
    "4": "자연과학",
    "5": "기술과학",
    "6": "예술",
    "7": "언어",
    "8": "문학",
    "9": "역사"
}

topic_order = [
    "총류",
    "철학",
    "종교",
    "사회과학",
    "자연과학",
    "기술과학",
    "예술",
    "언어",
    "문학",
    "역사"
]

def extract_district(file_path):
    file_name = os.path.basename(file_path)
    file_name = unicodedata.normalize("NFC", file_name)
    file_name = os.path.splitext(file_name)[0]
    parts = file_name.split("_")

    if len(parts) >= 3:
        return parts[1].strip()

    return None

def read_csv_auto(file_path):
    encodings = ["utf-8-sig", "utf-8", "cp949", "euc-kr"]

    for encoding in encodings:
        try:
            return pd.read_csv(
                file_path,
                encoding=encoding,
                low_memory=False
            )
        except UnicodeDecodeError:
            continue

    raise ValueError(f"파일을 읽을 수 없습니다: {file_path}")

topic_summary_list = []
classification_check_list = []
file_error_list = []

for file_path in loan_files:

    district = extract_district(file_path)

    if district is None:
        file_error_list.append({
            "파일명": os.path.basename(file_path),
            "오류": "자치구명 추출 실패"
        })
        continue

    try:
        df = read_csv_auto(file_path)
    except Exception as error:
        file_error_list.append({
            "파일명": os.path.basename(file_path),
            "오류": str(error)
        })
        continue

    df.columns = df.columns.astype(str).str.strip()

    required_columns = [
        "주제분류번호",
        "도서권수",
        "대출건수"
    ]

    missing_columns = [
        column
        for column in required_columns
        if column not in df.columns
    ]

    if missing_columns:
        file_error_list.append({
            "파일명": os.path.basename(file_path),
            "오류": f"필수 열 누락: {missing_columns}"
        })
        continue

    df["도서권수"] = pd.to_numeric(
        df["도서권수"],
        errors="coerce"
    ).fillna(0)

    df["대출건수"] = pd.to_numeric(
        df["대출건수"],
        errors="coerce"
    ).fillna(0)

    df.loc[df["도서권수"] < 0, "도서권수"] = 0
    df.loc[df["대출건수"] < 0, "대출건수"] = 0

    df["주제분류번호_원본"] = (
        df["주제분류번호"]
        .astype("string")
        .fillna("")
        .str.strip()
    )

    df["KDC번호_추출"] = (
        df["주제분류번호_원본"]
        .str.extract(
            r"([0-9]{3}(?:\.[0-9]+)?)",
            expand=False
        )
    )

    df["KDC앞자리"] = df["KDC번호_추출"].str[0]
    df["KDC주제"] = df["KDC앞자리"].map(kdc_map)

    valid_df = df[df["KDC주제"].notna()].copy()
    invalid_df = df[df["KDC주제"].isna()].copy()

    total_rows = len(df)
    total_books = df["도서권수"].sum()
    total_loans = df["대출건수"].sum()

    invalid_rows = len(invalid_df)
    invalid_books = invalid_df["도서권수"].sum()
    invalid_loans = invalid_df["대출건수"].sum()

    classification_check_list.append({
        "자치구": district,
        "전체행수": total_rows,
        "분류가능행수": len(valid_df),
        "미분류행수": invalid_rows,
        "전체도서권수": total_books,
        "미분류도서권수": invalid_books,
        "전체대출건수": total_loans,
        "미분류대출건수": invalid_loans,
        "행기준_미분류비율(%)": (
            invalid_rows / total_rows * 100
            if total_rows > 0
            else np.nan
        ),
        "도서권수기준_미분류비율(%)": (
            invalid_books / total_books * 100
            if total_books > 0
            else np.nan
        ),
        "대출건수기준_미분류비율(%)": (
            invalid_loans / total_loans * 100
            if total_loans > 0
            else np.nan
        )
    })

    grouped = (
        valid_df
        .groupby(
            "KDC주제",
            as_index=False,
            observed=False
        )
        .agg(
            주제별_도서권수=("도서권수", "sum"),
            주제별_대출건수=("대출건수", "sum")
        )
    )

    grouped["자치구"] = district
    topic_summary_list.append(grouped)

if len(topic_summary_list) == 0:
    raise ValueError("정상적으로 처리된 파일이 없습니다.")

topic_turnover = pd.concat(
    topic_summary_list,
    ignore_index=True
)

districts = sorted(
    topic_turnover["자치구"]
    .dropna()
    .unique()
)

full_index = pd.MultiIndex.from_product(
    [districts, topic_order],
    names=["자치구", "KDC주제"]
)

topic_turnover = (
    topic_turnover
    .set_index(["자치구", "KDC주제"])
    .reindex(full_index)
    .reset_index()
)

topic_turnover[
    ["주제별_도서권수", "주제별_대출건수"]
] = (
    topic_turnover[
        ["주제별_도서권수", "주제별_대출건수"]
    ]
    .fillna(0)
)
```

```python
topic_turnover["주제별_장서회전율"] = np.where(
    topic_turnover["주제별_도서권수"] > 0,
    topic_turnover["주제별_대출건수"]
    / topic_turnover["주제별_도서권수"],
    np.nan
)

topic_turnover["로그_주제별_장서회전율"] = np.log1p(
    topic_turnover["주제별_장서회전율"]
)

def minmax_by_topic(series):
    valid_series = series.dropna()

    if len(valid_series) == 0:
        return pd.Series(
            np.nan,
            index=series.index
        )

    min_value = valid_series.min()
    max_value = valid_series.max()

    if max_value == min_value:
        return pd.Series(
            np.where(
                series.notna(),
                50.0,
                np.nan
            ),
            index=series.index
        )

    return (
        (series - min_value)
        / (max_value - min_value)
        * 100
    )

topic_turnover["주제별_장서회전점수"] = (
    topic_turnover
    .groupby(
        "KDC주제",
        observed=False
    )["로그_주제별_장서회전율"]
    .transform(minmax_by_topic)
)

topic_turnover["KDC주제"] = pd.Categorical(
    topic_turnover["KDC주제"],
    categories=topic_order,
    ordered=True
)

topic_turnover = (
    topic_turnover
    .sort_values(["자치구", "KDC주제"])
    .reset_index(drop=True)
)

topic_turnover[
    [
        "주제별_장서회전율",
        "로그_주제별_장서회전율",
        "주제별_장서회전점수"
    ]
] = (
    topic_turnover[
        [
            "주제별_장서회전율",
            "로그_주제별_장서회전율",
            "주제별_장서회전점수"
        ]
    ]
    .round(4)
)

classification_check = pd.DataFrame(
    classification_check_list
)

classification_check = (
    classification_check
    .sort_values(
        "대출건수기준_미분류비율(%)",
        ascending=False
    )
    .reset_index(drop=True)
)

percentage_columns = [
    "행기준_미분류비율(%)",
    "도서권수기준_미분류비율(%)",
    "대출건수기준_미분류비율(%)"
]

classification_check[percentage_columns] = (
    classification_check[percentage_columns]
    .round(2)
)

file_error_result = pd.DataFrame(file_error_list)

display(topic_turnover)
display(classification_check)
display(file_error_result)
```

```python
output_path = (
    "/content/drive/MyDrive/ESAA YB 4조/*** 최종 ***/주제별_장서회전점수.csv"
)

topic_turnover.to_csv(
    output_path,
    index=False,
    encoding="utf-8-sig"
)

print("저장 완료:", output_path)
```

### 2-2. 주제선호도점수
해당 자치구의 해당 KDC 주제 대출건수 / 해당 자치구의 전체 도서 대출건수 × 100


```python
import pandas as pd

df = pd.read_csv(
    "/content/drive/MyDrive/ESAA YB 4조/*** 최종 ***/주제수요점수.csv",
    encoding="utf-8-sig"
)

display(df)
```

```python
#주제선호도점수
preference_rate = df.copy()

topic_columns = [
    "총류",
    "철학",
    "종교",
    "사회과학",
    "자연과학",
    "기술과학",
    "예술",
    "언어",
    "문학",
    "역사"
]

preference_score = df[["자치구"]].copy()

for column in topic_columns:
    min_value = df[column].min()
    max_value = df[column].max()

    if max_value == min_value:
        preference_score[f"{column}_선호도점수"] = 50.0
    else:
        preference_score[f"{column}_선호도점수"] = (
            (df[column] - min_value)
            / (max_value - min_value)
            * 100
        )

score_columns = [
    column
    for column in preference_score.columns
    if column != "자치구"
]

preference_score[score_columns] = (
    preference_score[score_columns]
    .round(4)
)

display(preference_score)
```

```python
preference_long = preference_score.melt(
    id_vars="자치구",
    var_name="KDC주제",
    value_name="주제선호도점수"
)

preference_long["KDC주제"] = (
    preference_long["KDC주제"]
    .str.replace("_선호도점수", "", regex=False)
)

preference_long = (
    preference_long
    .sort_values(["자치구", "KDC주제"])
    .reset_index(drop=True)
)

display(preference_long)
```

```python
output_path = (
    "/content/drive/MyDrive/ESAA YB 4조/*** 최종 ***/주제별_선호도점수.csv"
)

preference_long.to_csv(
    output_path,
    index=False,
    encoding="utf-8-sig"
)

print("저장 완료:", output_path)
```

```python
import pandas as pd
import re
from pathlib import Path

input_folder = Path(
    "/content/drive/MyDrive/ESAA YB 4조/*** 최종 ***/"
    "행정구역별 생활인구/자치구_매핑완료"
)

output_folder = input_folder / "생활인구점수_산출결과"
output_folder.mkdir(parents=True, exist_ok=True)
```

```python
file_list = sorted([
    item
    for item in input_folder.iterdir()
    if item.is_file() and item.suffix.lower() == ".csv"
])

print("찾은 파일 수:", len(file_list))

for file_path in file_list[:10]:
    print(file_path.name)
```

```python
import pandas as pd
import re

monthly_results = []
error_files = []

for file_path in file_list:

    file_name = file_path.name
    print(f"\n처리 중: {file_name}")

    try:
        df = pd.read_csv(
            file_path,
            encoding="utf-8-sig",
            dtype={
                "행정동코드": str,
                "자치구코드": str,
                "자치구명": str,
                "기준연월": str
            }
        )

        df["시간대구분"] = pd.to_numeric(
            df["시간대구분"],
            errors="coerce"
        )

        df["총생활인구수"] = pd.to_numeric(
            df["총생활인구수"],
            errors="coerce"
        )

        df = df.dropna(
            subset=[
                "기준일ID",
                "시간대구분",
                "총생활인구수",
                "자치구명"
            ]
        ).copy()

        # 도서관 이용 가능 시간대: 09시~19시
        df_time = df[
            df["시간대구분"].between(9, 19)
        ].copy()

        # 날짜·시간대·자치구별 행정동 생활인구 합산
        gu_hourly = (
            df_time
            .groupby(
                ["기준일ID", "시간대구분", "자치구명"],
                as_index=False
            )["총생활인구수"]
            .sum()
            .rename(
                columns={
                    "총생활인구수": "자치구_시간대생활인구"
                }
            )
        )

        # 날짜별 자치구 평균생활인구
        gu_daily = (
            gu_hourly
            .groupby(
                ["기준일ID", "자치구명"],
                as_index=False
            )["자치구_시간대생활인구"]
            .mean()
            .rename(
                columns={
                    "자치구_시간대생활인구": "일평균생활인구"
                }
            )
        )

        year_month = re.search(
            r"(\d{6})",
            file_name
        ).group(1)

        # 월별 자치구 평균생활인구
        gu_monthly = (
            gu_daily
            .groupby(
                "자치구명",
                as_index=False
            )["일평균생활인구"]
            .mean()
            .rename(
                columns={
                    "일평균생활인구": "월평균생활인구"
                }
            )
        )

        gu_monthly["기준연월"] = year_month

        gu_monthly = gu_monthly[
            ["기준연월", "자치구명", "월평균생활인구"]
        ]

        monthly_results.append(gu_monthly)

        print(
            f"완료: {year_month} | "
            f"자치구 수: {gu_monthly['자치구명'].nunique()}"
        )

    except Exception as e:
        error_files.append({
            "파일명": file_name,
            "오류내용": str(e)
        })

        print("오류:", e)
```

```python
# Min-Max 정규화
def minmax_100(series):
    min_val = series.min()
    max_val = series.max()

    if max_val == min_val:
        return series * 0

    return (series - min_val) / (max_val - min_val) * 100

pop["인구증가율점수"] = minmax_100(pop["인구증가율"])
```

```python
import pandas as pd

current_gu = "종로구"

current_potential = (
    potential[
        potential["자치구"] == current_gu
    ][
        ["KDC주제", "Potential Score"]
    ]
    .rename(
        columns={
            "Potential Score": "현재_Potential_Score"
        }
    )
)

transfer_score = (
    potential[
        potential["자치구"] != current_gu
    ][
        ["자치구", "KDC주제", "Potential Score"]
    ]
    .rename(
        columns={
            "자치구": "후보_자치구",
            "Potential Score": "후보_Potential_Score"
        }
    )
    .merge(
        current_potential,
        on="KDC주제",
        how="left"
    )
)

transfer_score["Transfer Score"] = (
    transfer_score["후보_Potential_Score"]
    - transfer_score["현재_Potential_Score"]
)

transfer_score = (
    transfer_score
    .sort_values(
        ["KDC주제", "Transfer Score"],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)

display(transfer_score)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report
)
from sklearn.preprocessing import label_binarize
import pandas as pd
import numpy as np

models = {
    "Random Forest": rf_model,
    "XGBoost": xgb_model,
    "LightGBM": lgbm_model,
    "Logistic Regression": lr_model
}

evaluation_results = []

for model_name, model in models.items():

    model.fit(X_train_scaled, y_train)

    y_pred = model.predict(X_test_scaled)
    y_prob = model.predict_proba(X_test_scaled)

    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(
        y_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    try:
        roc_auc = roc_auc_score(
            y_test,
            y_prob,
            multi_class="ovr",
            average="macro",
            labels=model.classes_
        )
    except ValueError:
        roc_auc = np.nan

    evaluation_results.append({
        "모델": model_name,
        "정확도": accuracy,
        "정밀도_Macro": precision,
        "재현율_Macro": recall,
        "F1_Score_Macro": f1,
        "ROC_AUC_Macro_OVR": roc_auc
    })

evaluation_df = pd.DataFrame(evaluation_results)

numeric_columns = [
    "정확도",
    "정밀도_Macro",
    "재현율_Macro",
    "F1_Score_Macro",
    "ROC_AUC_Macro_OVR"
]

evaluation_df[numeric_columns] = (
    evaluation_df[numeric_columns]
    .round(4)
)

display(
    evaluation_df.sort_values(
        "F1_Score_Macro",
        ascending=False
    )
)
```

##5. 클래스별 ROC 곡선

###5-1. 고정 테스트 셋

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import label_binarize
from sklearn.metrics import roc_curve, auc

class_labels = [0, 1, 2]
class_names = ["하", "중", "상"]

y_test_bin = label_binarize(
    y_test,
    classes=class_labels
)

for model_name, model in models.items():

    model.fit(X_train_scaled, y_train)

    y_prob = model.predict_proba(X_test_scaled)

    plt.figure(figsize=(7, 5))

    valid_curve_count = 0

    for i, class_name in enumerate(class_names):

        if len(np.unique(y_test_bin[:, i])) < 2:
            print(
                f"{model_name} - {class_name}: "
                "테스트셋에 양성 또는 음성 표본이 없어 ROC 계산 불가"
            )
            continue

        fpr, tpr, _ = roc_curve(
            y_test_bin[:, i],
            y_prob[:, i]
        )

        class_auc = auc(fpr, tpr)

        plt.plot(
            fpr,
            tpr,
            label=f"{class_name} vs Rest (AUC={class_auc:.3f})"
        )

        valid_curve_count += 1

    plt.plot(
        [0, 1],
        [0, 1],
        linestyle="--",
        label="Random"
    )

    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title(f"{model_name} ROC Curve - Fixed Test Set")
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()

    if valid_curve_count == 0:
        print(f"{model_name}: 표시할 수 있는 ROC 곡선이 없습니다.")
```

```python
df_priority["장기미대출비율"] = (
    df_priority["장기미대출권수"] / df_priority["총소장권수"]
) * 100

df_priority["장서회전율"] = (
    df_priority["연간대출건수"] / df_priority["총소장권수"]
)
min_t = df_priority["장서회전율"].min()
max_t = df_priority["장서회전율"].max()
df_priority["장서회전율정규화"] = (df_priority["장서회전율"] - min_t) / (max_t - min_t) * 100

df_priority = df_priority.merge(pop_calc[["자치구", "인구증가율"]], on="자치구", how="left")
df_priority = df_priority.merge(df_households_per_gu, on="자치구", how="left")

avg_household = df_households_per_gu["총세대수"].mean()
df_priority["미래대출예측량"] = (
    df_priority["연간대출건수"]
    * (1 + df_priority["인구증가율"] / 100)
    * (df_priority["총세대수"] / avg_household)
)

min_f = df_priority["미래대출예측량"].min()
max_f = df_priority["미래대출예측량"].max()
df_priority["미래대출지수정규화"] = (
    (df_priority["미래대출예측량"] - min_f) / (max_f - min_f)
) * 100

df_priority["PriorityScore"] = (
    0.45 * df_priority["장기미대출비율"]
    + 0.30 * (100 - df_priority["장서회전율정규화"])
    + 0.25 * (100 - df_priority["미래대출지수정규화"])
)
```
